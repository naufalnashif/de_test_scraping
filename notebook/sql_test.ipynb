```python
import os
import sys

# Menambahkan root directory ke sys.path
# Ini akan naik satu level dari folder 'notebooks' ke root project
module_path = os.path.abspath(os.path.join('..'))
if module_path not in sys.path:
    sys.path.append(module_path)

import pandas as pd
from sqlalchemy import text
from app.database import DatabaseConnection
```

```python
def run_sql_etl(sql_file_name, table_name):
    """
    Membaca file SQL, mengeksekusi ke database, dan mengembalikan DataFrame hasil.
    """
    engine = DatabaseConnection.get_engine()
    sql_path = os.path.join('..', 'scripts', 'sql', sql_file_name)
    
    try:
        # 1. Baca query dari file
        with open(sql_path, 'r') as f:
            query = f.read()
            
        # 2. Jalankan perintah SQL (Drop & Create)
        with engine.begin() as conn:
            # Opsional: Drop table lama jika ingin refresh data total
            conn.execute(text(f"DROP TABLE IF EXISTS transformed.{table_name}"))
            conn.execute(text(query))
            print(f"✓ ETL Success: Table transformed.{table_name} created.")

        # 3. Ambil data untuk print out ke notebook
        df = pd.read_sql(f"SELECT * FROM transformed.{table_name}", engine)
        return df

    except Exception as e:
        print(f"❌ Error saat menjalankan {sql_file_name}: {e}")
        return None
```

```python
# Jalankan Product Specs
df_specs = run_sql_etl('/Users/naufalnashif/Desktop/de_test_danone/scripts/dm_product_specs.sql', 'dm_product_specs')
if df_specs is not None:
    display(df_specs)
```
